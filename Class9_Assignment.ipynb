```python
import pandas as pd
import numpy as np
```

Q1:  Create a DataFrame containing 8 students with columns such as Name, Marks, 
Age, and City. Include some missing values.

```python
score = np.array([ ["Amar", 74, 18, "Baripada"], ["Biswa", None, 17, "Toronto"], ["Chirag", 89, 17, "Bhubaneswar"], ["Dhanush", 93, None, "Ooty"], ["Eeshant", 96, 19, "Telengana"], ["Biswa", None, 17, "Toronto"], ["Fanhi", 84, 17, None], ["Gaitonde", 90, 18, "Vimanagar"], ["Harsh", 67, 18, "Rourkela"] ])

report = pd.DataFrame(score, columns=['Name', 'Marks', 'Age', 'City'])

report.head(8)
```

Q2: Find the number of missing values in each column using isnull()

```python
report[report.isnull().any(axis = 1)].shape[0]
```

Q3: Display all rows that contain at least one missing value.

```python
report[report.isnull().any(axis=1)]
```

Q4: Remove all rows containing missing values using dropna().

```python
report.dropna()
```

Q5: Remove only the rows where the Marks column contains missing values.

```python
report.dropna(subset=["Marks"])
```

Q6: Fill missing values in the Marks column using the mean of the column.

```python
report["Marks"] = report["Marks"].fillna(report["Marks"].mean().round(2))

print(report)
```

Q7: Fill missing values in the Age column using the median of the column.

```python
report["Age"] = report["Age"].fillna(report["Age"].median())

report.head(8)
```

Q8: Replace missing values in the City column with "Unknown"

```python
report["City"] = report["Age"].fillna("Unknown")

report.head()
```

Q9: Create a DataFrame containing duplicate records and identify the duplicate rows 
using duplicated().

```python
report[report.duplicated()]
```

Q10: Remove duplicate rows using drop_duplicates().

```python
report.drop_duplicates()
```
