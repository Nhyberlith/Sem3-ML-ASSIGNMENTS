Q2 : You intercept a secret code where words are separated by hyphens (-). To decode it, you must reverse the order of the words, convert every letter to UPPERCASE, and replace the hyphens with spaces.

```python
SecretCode = "charlie-bravo-alpha"

def DECODER(code) :

    lst = code.split("-")
    lst[:] = lst[::-1]
    print(" ".join(lst).upper())


#input
DECODER(SecretCode)
```

*Q3 : You have received a raw string of names from a database exported as a semicolon-separated list. The data is messy, full of random spaces, and has inconsistent casing. Write a function to clean this string so that every name is properly Title Cased, extra spaces are removed, and the names are joined into a single string separated by a comma and a space.

```python
raw_names = " john doe ; jAnE sMiTh; ALAN tUrInG "

def dbCleaner(data) :

    lst = data.split(";")
    
    for names in lst :
        print(names.lstrip().rstrip().lower().title(), end=",\n")
    

#input
dbCleaner(raw_names)
```

Q5 : Determine if a given sentence is a palindrome. To do this correctly using only string methods, remove all spaces and non-alphanumeric characters, convert everything to lowercase, and check whether it reads the same forwards and backwards.

```python
case_1 = "Was it a car or a cat I saw?" 
case_2 = "Python is fun" 

def isPalindrome(text) :  

    s = " ".join(text).lower()
    lst = s.split()

    for element in lst :

        if element.isalnum() == False :
            lst.remove(element)
    

    i = 0
    j = len(lst) - 1

    while i < j :

        if lst[i] != lst[j] :
            return False
        
        i+=1
        j-=1
        
    return True


    

#input
isPalindrome(case_2)
```

Q9 : Write a function to check if two given strings are anagrams of each other — they contain the exact same letters in the exact same frequencies, just rearranged. Ignore any spaces and treat uppercase and lowercase letters as the same.
Note: Solve it using string methods and basic built-in functions (like sorted()).

```python
str1 = "Dormitory"
str2 = "Dirty Room"

def isAnagram(a,b) :

    a = ''.join(a.lower().split())
    b = ''.join(b.lower().split())

    print( "".join(sorted(a)) == "".join(sorted(b)) )


#input
isAnagram(str1, str2)
```

Q10 : You are building a CRM system and user-submitted phone numbers are a mess — they contain spaces, parentheses, periods, and hyphens. Write a function that takes a raw phone number string and cleans it by removing all non-digit characters. Once cleaned, validate the number:
• If the cleaned string contains exactly 10 digits, format and return it as "(XXX) XXX-XXXX".
• If it contains anything other than 10 digits, return "Invalid Number".
Note: Case 1 fails because after removing symbols it has 11 digits, due to the leading “1”.

```python
raw_num1 = " +1 (555) 123.4567 "
raw_num2 = "555-987-6543"

def ContactNoFormatter(num) :
    s = ''.join(num.split())
    lst = list(s)

    for element in lst :
        if not element.isalnum() :
            lst.remove(element)

    if len(lst) == 10 :
        lst.insert(0, "(")
        lst.insert(4, ") ")
        lst.insert(8, "-")
        print("".join(lst))
    else :
        print("Invalid Number")


#input
ContactNoFormatter(raw_num1)
```

Q11 : An inventory system requires product codes (SKUs) to strictly match this format after being cleaned up: exactly 3 uppercase letters, a single hyphen, and exactly 4 digits (e.g. "ABC-1234"). Write a function that receives a messy SKU string. First clean it by stripping leading/trailing spaces and converting it to uppercase. Then validate it using string methods. Return "Valid: [Cleaned_SKU]" if it passes, or "Invalid SKU" if it fails any formatting rule.
Note: Use isalpha(), isdigit() etc. — no re module. Case 2 fails (2 letters, 5 digits); Case 3 fails (special character).

```python
sku1 = " xYz-9876 " 
sku2 = "AB-12345"
sku3 = "A@C-1234"


def SKUinspector(code) :
    
    s = code.strip().upper()
    
    check = s.split("-")

    if check[0] == s : 
        return "Invalid SKU"

    if ( len(check[0])== 3 and check[0].isalpha() ) and ( len(check[1])==4 and check[1].isdigit() ) :
        return "Valid: " + s
    else :
        return "Invalid SKU"



#input
SKUinspector(sku3)
```
