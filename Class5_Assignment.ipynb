Q1. A class has 4 students and 3 subjects. Their marks are given below. The examiner announces a
subject-wise bonus of 5, 2 and 8 marks respectively. Add the bonus to every student in one line using
broadcasting, then cap any mark above 100 back to 100 and report how many marks had to be capped.

```python
marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])

bonus = np.array([5, 2, 8])

newMarks = marks + bonus    #broadcasting

newMarksCapped = np.where(newMarks > 100 , 100, newMarks)     
cappedCount = np.sum(newMarks > 100)    #summing the number of cases which return true, not the sum of values

print(newMarksCapped)
print(cappedCount)
```

Q5. Given a 1D array of 12 monthly sales figures, reshape it into 4 quarters of 3 months each and find
which quarter (index) had the highest average sales, along with that average.

```python
sales = np.array([120, 135, 98, 160, 175, 190, 88, 92, 105, 210, 198, 230])

X = sales.reshape(4,-1)

avgSalesPerQuart = X.mean(axis=1)       #axis 1 means (evaluation) per row. axis 0 means the same for column
highestAvg = np.max(avgSalesPerQuart)
bestQuartIndex = np.argmax(avgSalesPerQuart)        #argmax returs index of the max value 

print(highestAvg)
print(bestQuartIndex)
```

Q8. For the matrix below, scale every column into the 0 to 1 range using column-wise min-max
normalization (x - col_min) / (col_max - col_min). Use broadcasting only — no loops. Verify that
each column of the result has a minimum of 0 and a maximum of 1.

```python
X = np.array([[ 50, 1200,  3],
              [ 80, 1500,  9],
              [ 65,  900,  5],
              [ 95, 2000,  7]])

col_min = X.min(axis=0)
col_max = X.max(axis=0)

Result = (X - col_min) / (col_max - col_min)

print(Result.round(2))
#verification
print("min of each col", Result.min(axis=0).astype(int))
print("max of each col", Result.max(axis=0).astype(int))
```

Q15. Apply min-max normalization to scale any array into the 0 to 1 range using (x - min) / (max -
min).

```python
x = np.array([10, 20, 30, 40, 50])

x_min = x.min()
x_max = x.max()

normalized_x = (x - x_min) / (x_max - x_min)

print(x)
print(normalized_x)
```
