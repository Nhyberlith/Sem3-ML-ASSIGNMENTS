```python
marks = np.array([[70, 65, 80],
                  [55, 90, 32],
                  [88, 76, 91],
                  [40, 52, 45],
                  [60, 58, 66]])
attendance = np.array([88, 1, 105, 76, 45, 1, 92, 120, 67, 30])

capped = np.minimum(attendance, 100)
repaired = np.where(capped == 1, capped[capped != 1].mean(), capped)
print(repaired)

labels = np.where(repaired >= 85, 'Regular', np.where(repaired >= 60, 'Average', 'Short'))

short_idx = np.where(labels == 'Short')[0]
print(short_idx, repaired[short_idx])

rows, cols = np.where(marks < 50)
print("rows:", rows, "\ncols:", cols)

print(marks[rows, cols])

changed_count = np.sum(marks < 35)
marks = np.where(marks < 35, 35, marks)
print(changed_count)

print((marks < 60).sum(axis=1).argmax())
```
