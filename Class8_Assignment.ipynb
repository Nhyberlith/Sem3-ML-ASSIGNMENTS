B5: Make a column called age_group. Everyone starts as 'adult'. Anyone under 18 becomes 'child'. Then count
how many of each.

```python


t['age_group'] = 'adult'

t.loc[ t['Age'].isnull(), 'age_group' ] = 'data not found'

t.loc[ t['Age'] < 18, 'age_group'] = 'child'

t['age_group'].value_counts()


# by default we if ignore the 2nd line of code, NaN data is converted to adults. Python. When we run the evaluation of t['Age'] < 18, NaN < 18 is treated as false.
```
