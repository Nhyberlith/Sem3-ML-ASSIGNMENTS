Q2. Prices arrive as one messy string: " 120, 45 , 300 ,80, 95 ". Convert it into a list of integers, then print the total bill, the highest price, and the average price with 2 decimal places (use an f-string with :.2f).

```python
bill = " 120, 45 , 300 ,80, 95 "

FinalBill = bill.split(",")

FinalBill = [ int(x) for x in FinalBill ]

total = 0
Highest = FinalBill[0]

for price in FinalBill :

    total = total + price

    if price > Highest :
        Highest = price



print( "Total :", total)
print("Highest :", Highest)
print("Average :", total/len(FinalBill))
```

Q3. Validate a username entered by the user (e.g. " Rahul_123 "). Rules:
• clean the spaces and convert to lowercase
• must be alphanumeric (letters and digits only — no symbols)
• must start with an alphabet
• length must be between 5 and 12 characters
Print Valid username or Invalid username. ("Rahul_123" is invalid because of the underscore; "rahul123" is valid.)

```python
userName =  " Rahul_123 "

name = list(userName.lower().strip())

for element in name :

    if not element.isalnum() :
        name.remove(element)

print("".join(name))
```

Q4. For a given sentence, print: (a) the total number of words, (b) how many words end with a vowel, and (c) the longest word in the sentence.

```python
sentence = "the punishment given by a judge to somebody who has been found guilty of a crime"
lst = sentence.split()

totalWords = 0
endWithVowel = 0
longestWordLength = len(lst[0])
longestWord = ""

for words in lst :

    totalWords = totalWords + len(words)
    
    if words[-1] == "a" or words[-1] == "e" or words[-1] == "i" or words[-1] == "o" or words[-1] == "u" :
        endWithVowel += 1

    if len(words) > longestWordLength :
        longestWord = words
        longestWordLength = len(words)



print(totalWords)
print(endWithVowel)
print(longestWord)
```

Q5. Count the frequency of each character in a string, ignoring spaces and treating uppercase and lowercase as the same letter. Example: "Aba c" → {'a': 2, 'b': 1, 'c': 1}.

```python
string = "Aba c"
d = {}

lst = list(string.lower())

for char in lst :

    if char.isalpha() :
        
        if char in d :
            d[char] = d[char] + 1
        else :
            d[char] = 1

print(d)
```

Q7. Invert the dictionary below so that values become keys. If two days share the same value, store both days in a list:

```python
sales = {"mon": 200, "tue": 350, "wed": 200, "sun": 200}
invert = {}

for key in sales:
    value = key
    newKey = sales[key]

    if newKey in invert:
        invert[newKey].append(value)
    else:
        invert[newKey] = [value]

print(invert)
```

Q8. Marks of two tests are stored in two dictionaries. Merge them into one dictionary — if a student appears in
both, sum the marks:

```python
test1 = {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}

for student in test2 :

    if student in test1 :
        test1[student] = test1[student] + test2[student]

    else :
        test1[student] = test2[student]

print(test1)
```

Q9. Sort the employees below by salary (highest first) using sorted() with key=lambda, and print each line
using an f-string in the format shown:

```python
employees = [("Ravi", 55000), ("Sita", 72000), ("Amit", 48000)]

employees.sort(key=lambda x: x[1], reverse=True)

for data in employees :

    print( data[0], "earns Rs.", data[1] )
```
